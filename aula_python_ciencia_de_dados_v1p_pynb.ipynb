:### 1.5 Orientação a Objetos (visão geral - Indo além)

```python
class Aluno:
    def __init__(self, nome, notas):
        self.nome = nome
        self.notas = notas

    def media(self):
        return sum(self.notas) / len(self.notas)

    def status(self):
        return "Aprovado" if self.media() >= 7 else "Reprovado"

a1 = Aluno("Bruno", [8, 7.5, 9])
print(f"{a1.nome}: média {a1.media():.1f} -> {a1.status()}")
```

Resumindo o conceito de classe:

1. **`class Aluno:`** — Define o "molde" da classe, que servirá para criar objetos com atributos (características) e métodos (ações).

2. **`__init__(self, nome, notas)`** — É o construtor, chamado automaticamente ao criar um objeto. Ele inicializa os atributos `nome` e `notas` de cada aluno.

3. **`media(self)`** — Método que calcula a média das notas: soma todas (`sum`) e divide pela quantidade (`len`).

4. **`status(self)`** — Método que usa a média calculada para decidir o resultado: `"Aprovado"` se a média for ≥ 7, senão `"Reprovado"`.

5. **`a1 = Aluno("Bruno", [8, 7.5, 9])`** — Cria uma *instância* (objeto real) da classe, passando os dados que serão usados pelos métodos `a1.media()` e `a1.status()` na hora de exibir o resultado.

0BS: self é a forma como cada objeto se refere a si mesmo dentro da classe

# 2. NumPy 🔢

**NumPy** (Numerical Python) é a base de todo o ecossistema científico do Python. Ele fornece o `array`, uma estrutura muito mais rápida e eficiente que listas para cálculos numéricos.


### 2.1 Criando arrays

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4, 5])
arr2d = np.array([[1, 2, 3],
                  [4, 5, 6]])

print("Array 1D:", arr1)
print("Array 2D:\n", arr2d)
print("Shape do array 2D:", arr2d.shape)

# Arrays especiais
zeros = np.zeros((2, 3))
uns = np.ones((3, 3))
identidade = np.eye(3)
sequencia = np.arange(0, 10, 2)     # de 0 a 10, pulando de 2
linspace = np.linspace(0, 1, 5)     # 5 números igualmente espaçados entre 0 e 1

print("\nzeros:\n", zeros)
print("arange:", sequencia)
print("linspace:", linspace)
```

### 2.2 Operações vetorizadas (o grande poder do NumPy)

```python
precos = np.array([100, 250, 80, 400, 150])
desconto = 0.10

# Sem NumPy você precisaria de um for. Com NumPy, é direto:
precos_com_desconto = precos * (1 - desconto)
print(precos_com_desconto)

print("Média:", precos.mean())
print("Desvio padrão:", precos.std())
print("Coeficiente de variação:", precos.std()/precos.mean()*100)
print("Máximo:", precos.max(), "| Mínimo:", precos.min())
print("Soma:", precos.sum())
```

### 2.3 Indexação, slicing e máscaras booleanas

```python
notas_np = np.array([6.5, 8.0, 9.5, 4.0, 7.2, 10.0])

print("Primeiros 3:", notas_np[:3])
print("Últimos 2:", notas_np[-2:])

# Máscara booleana: filtro poderosíssimo, muito usado em pandas também
mask = notas_np >= 7
print("Máscara:", mask)
print("Aprovados:", notas_np[mask])

#Alternativas
print(f"Aprovados:", notas_np[mask])
print(f"Aprovados: {notas_np[mask]}")

#OBS: Note a diferença das duas abordagens com f e sem f, trabalhando com as variáveis no colchete
#jeito correto: f-string print(f"texto {variavel}")
```

### 2.4 Álgebra linear básica

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[2, 0],
              [1, 2]])

print("Soma:\n", A + B)
print("Multiplicação de matrizes:\n", A @ B)
print("Transposta de A:\n", A.T)
print("Determinante de A:", np.linalg.det(A))
print("Inversa de A:\n", np.linalg.inv(A))
```

matriz $A = \begin{pmatrix} a & b \\ c & d \end{pmatrix}$

Calcular o determinante:
Multiplique os elementos da diagonal principal e subtraia a multiplicação da diagonal secundária.

$$\det(A) = (a \cdot d) - (b \cdot c)$$

O resultado da inversão de uma matriz 2x2 é dado por:

$$A^{-1} = \frac{1}{\det(A)} \cdot \begin{pmatrix} d & -b \\ -c & a \end{pmatrix}$$

# 3. Pandas — DataFrames para resolver tarefas complexas 🐼

**pandas** é a biblioteca de manipulação de dados mais usada em Python. O objeto central é o **DataFrame**, uma tabela com linhas e colunas (parecido com uma planilha ou um `data.frame` do R).


### 3.1 Criando um DataFrame

```python
import pandas as pd
import numpy as np

np.random.seed(42)
dados = {
    "aluno": [f"Aluno_{i}" for i in range(1, 21)],
    "curso": np.random.choice(["Python", "R", "SQL", "Power BI"], 20),
    "nota": np.random.uniform(3, 10, 20).round(1),
    "horas_estudo": np.random.randint(5, 60, 20),
    "estado": np.random.choice(["RJ", "SP", "MG", "RS"], 20),
}

df = pd.DataFrame(dados)
df.head()
display(df) # mostra o df inteiro no colab
```

### 3.2 Explorando o DataFrame

```python
print(df.shape)          # (linhas, colunas)
print(df.dtypes)         # tipos de cada coluna
df.info()
df.describe()             # estatísticas descritivas das colunas numéricas
```

### 3.3 Seleção e filtro de dados

```python
# Selecionar colunas
print(df["nota"].head())
print(df[["aluno", "nota"]].head())

# Filtrar linhas (equivalente ao filter() do dplyr)
aprovados = df[df["nota"] >= 7]
print(aprovados.shape)

# Filtros combinados
filtro = (df["nota"] >= 7) & (df["curso"] == "Python")
df[filtro]
```

### 3.4 Criando e transformando colunas

```python
df["status"] = np.where(df["nota"] >= 7, "Aprovado", "Reprovado")
df["nota_normalizada"] = (df["nota"] - df["nota"].min()) / (df["nota"].max() - df["nota"].min())
df.head()
```

### 3.5 Agrupamentos (groupby) — o `SUMMARIZE` do pandas

```python
# Média de nota por curso
resumo_curso = df.groupby("curso")["nota"].agg(["mean", "std", "count"]).round(2)
resumo_curso.columns = ["media", "desvio_padrao", "n_alunos"]
resumo_curso.sort_values("media", ascending=False)
```

```python
# Tabela dinâmica (pivot_table) — cruzando curso x estado
pivot = df.pivot_table(index="curso", columns="estado", values="nota", aggfunc="mean").round(1)
pivot
```

```python
# Tabela dinâmica (pivot_table) — cruzando curso x status
pivot = df.pivot_table(index="curso", columns="status", values="nota", aggfunc="mean").round(1)
pivot
```

### 3.6 Valores ausentes e duplicados

```python
df_com_na = df.copy()
df_com_na.loc[2, "nota"] = np.nan #altera um valor específico do DataFrame: pega a linha de índice 2 e a coluna "nota", e substitui o valor que estava lá por np.nan
df_com_na.loc[5, "nota"] = np.nan

print("Valores nulos por coluna:")
print(df_com_na.isna().sum())

# Estratégias comuns
df_preenchido = df_com_na.fillna(df_com_na["nota"].mean())      # preencher com a média
df_sem_na = df_com_na.dropna(subset=["nota"])                    # remover linhas com NA
print("\nLinhas restantes após dropna:", df_sem_na.shape[0])
print("\nLinhas antes da remoção:", df.shape[0])
print("\nLinhas linhas removidas:", df.shape[0]  - df_sem_na.shape[0])
```

### 3.7 merge / join (juntar tabelas)

```python
df
cursos_info = pd.DataFrame({
    "curso": ["Python", "R", "SQL", "Power BI"],
    "carga_horaria": [40, 40, 20, 15],
})

df_join = df.merge(cursos_info, on="curso", how="left")
df_join.head()
```

Regra prática para lembrar:

- left = "quero manter tudo da minha tabela principal, mesmo que não ache correspondência"
- right = "quero manter tudo da tabela que estou juntando, mesmo que não ache correspondência"

OBS: Na dúvida, left é o mais usado no dia a dia, porque geralmente você parte de uma tabela "principal" (ex: alunos) e só quer enriquecer com informações de outra (ex: dados do curso) — sem perder nenhum aluno no processo.

# 8. Machine Learning: Regressão Linear 📐

Regressão Linear é o algoritmo mais fundamental de ML supervisionado: modela a relação **linear** entre uma variável dependente (`y`) e uma ou mais variáveis independentes (`X`).

$$ y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \epsilon $$


### 8.1 Preparando os dados

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Vamos prever 'nota' a partir de 'horas_estudo'
X = df[["horas_estudo"]]
y = df["nota"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
print("Treino:", X_train.shape, "| Teste:", X_test.shape)
```

### 8.2 Treinando o modelo

```python
modelo_lr = LinearRegression()
modelo_lr.fit(X_train, y_train)

print(f"Coeficiente (inclinação): {modelo_lr.coef_[0]:.4f}")
print(f"Intercepto: {modelo_lr.intercept_:.4f}")
print(f"\nInterpretação: cada hora extra de estudo aumenta a nota prevista em {modelo_lr.coef_[0]:.3f} pontos")
```

### 8.3 Avaliando o modelo

```python
y_pred = modelo_lr.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.2f}")
print(f"R²: {r2:.2f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(X_test, y_test, color="#264653", label="Valores reais")
plt.plot(X_test, y_pred, color="#E76F51", linewidth=2, label="Reta prevista")
plt.xlabel("Horas de estudo")
plt.ylabel("Nota")
plt.title("Regressão Linear: Horas de estudo x Nota")
plt.legend()
plt.show()
```

> 💡 Com poucos dados sintéticos o R² pode ser baixo — isso é esperado neste exemplo didático. Em um caso real, com mais observações e variáveis relevantes, o modelo tende a generalizar melhor.

### 🔧 Exercício 7
Adicione a coluna `nota_normalizada` como uma segunda variável explicativa (regressão linear múltipla) e compare o R² com o modelo simples.


```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import pandas as pd
import numpy as np

# --- Modelo simples ---
X1 = df[["horas_estudo"]]
Xtr1, Xte1, ytr1, yte1 = train_test_split(X1, y, test_size=0.25, random_state=42)
m1 = LinearRegression().fit(Xtr1, ytr1)
r2_1 = r2_score(yte1, m1.predict(Xte1))

# --- Modelo com vazamento (didático) ---
X2 = df[["horas_estudo", "nota_normalizada"]]
Xtr2, Xte2, ytr2, yte2 = train_test_split(X2, y, test_size=0.25, random_state=42)
m2 = LinearRegression().fit(Xtr2, ytr2)
r2_2 = r2_score(yte2, m2.predict(Xte2))

# --- Modelo corrigido (curso via one-hot) ---
X3 = pd.get_dummies(df[["horas_estudo", "curso"]], columns=["curso"], drop_first=True)
Xtr3, Xte3, ytr3, yte3 = train_test_split(X3, y, test_size=0.25, random_state=42)
m3 = LinearRegression().fit(Xtr3, ytr3)
r2_3 = r2_score(yte3, m3.predict(Xte3))

print(f"1) Simples (horas_estudo):              R² = {r2_1:.4f}")
print(f"2) Com vazamento (+ nota_normalizada):   R² = {r2_2:.4f}  <- suspeito!")
print(f"3) Corrigido (+ curso):                  R² = {r2_3:.4f}")
```

# 9. K-Nearest Neighbors (KNN) 👥

KNN classifica (ou prevê) um ponto novo olhando para os **K vizinhos mais próximos** dele no espaço de variáveis, e usando a classe majoritária (ou a média, em regressão) entre eles.


### 9.1 Preparando dados de classificação

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

X = df[["nota", "horas_estudo"]]
y = df["status"]   # Aprovado / Reprovado

# KNN é sensível à escala das variáveis -> sempre padronizar!
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.25, random_state=42, stratify=y)
```

### 9.2 Treinando e avaliando o KNN

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)
print("Acurácia:", accuracy_score(y_test, y_pred).round(2))
print("\n", classification_report(y_test, y_pred, zero_division=0))
```

### 9.3 Escolhendo o melhor K (método do cotovelo)

```python
import matplotlib.pyplot as plt

erros = []
for k in range(1, 15):
    knn_k = KNeighborsClassifier(n_neighbors=k)
    knn_k.fit(X_train, y_train)
    pred_k = knn_k.predict(X_test)
    erros.append(1 - accuracy_score(y_test, pred_k))

plt.figure(figsize=(8, 5))
plt.plot(range(1, 15), erros, marker="o", color="#E76F51")
plt.xlabel("K (número de vizinhos)")
plt.ylabel("Taxa de erro")
plt.title("Escolha do melhor K")
plt.grid(alpha=0.3)
plt.show()
```

# 10. K-Means Clustering 🎯

K-Means é um algoritmo de **aprendizado não supervisionado**: ele agrupa observações parecidas em **K clusters**, sem usar rótulos (labels) — ótimo para segmentação de clientes, alunos, produtos, etc.


### 10.1 Preparando os dados

```python
from sklearn.cluster import KMeans

X_cluster = df[["nota", "horas_estudo"]]
X_cluster_scaled = StandardScaler().fit_transform(X_cluster)
```

### 10.2 Escolhendo o número de clusters (método do cotovelo)

```python
inercia = []
K_range = range(1, 8)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_cluster_scaled)
    inercia.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inercia, marker="o", color="#2A9D8F")
plt.xlabel("Número de clusters (K)")
plt.ylabel("Inércia")
plt.title("Método do Cotovelo")
plt.grid(alpha=0.3)
plt.show()
```

### 10.3 Treinando o K-Means e visualizando os clusters

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X_cluster_scaled)

plt.figure(figsize=(8, 5))
scatter = plt.scatter(df["horas_estudo"], df["nota"], c=df["cluster"], cmap="viridis", s=80, edgecolor="white")
plt.xlabel("Horas de estudo")
plt.ylabel("Nota")
plt.title("Clusters de alunos (K-Means)")
plt.colorbar(scatter, label="Cluster")
plt.show()

df.groupby("cluster")[["nota", "horas_estudo"]].mean().round(2)
```

> 💡 **Aplicação real:** é assim que empresas segmentam clientes por comportamento de compra, engajamento ou perfil demográfico — por exemplo, agrupar alunos da CECD por "engajamento" (horas assistidas x nota em quizzes) para campanhas de marketing direcionadas.

Uma Árvore de Decisão faz perguntas de "sim/não" (ou "maior/menor que") em sequência para chegar a uma previsão. É um modelo fácil de interpretar e visualizar — cada divisão (split) busca separar melhor as classes.

# 13. Processamento de Linguagem Natural (NLP) 💬

NLP é a área de ML voltada para trabalhar com **texto**: comentários, avaliações, mensagens, transcrições de vídeo, etc. Vamos ver o pipeline básico: limpeza → tokenização → remoção de stopwords → vetorização → análise de sentimento.


### 13.1 Dados de exemplo (comentários fictícios de alunos)

```python
comentarios = [
    "Curso excelente, aprendi muito sobre Python e ciência de dados!",
    "Achei o conteúdo confuso e mal explicado, não recomendo.",
    "Professor muito didático, as aulas de estatística ficaram claras.",
    "Material bom, mas o suporte demora muito para responder.",
    "Adorei o curso de Machine Learning, superou minhas expectativas!",
    "Preço alto para o conteúdo oferecido, esperava mais.",
    "A comunidade é ótima, sempre tem alguém disposto a ajudar.",
    "Vídeos com áudio ruim atrapalham o aprendizado.",
]

df_comentarios = pd.DataFrame({"comentario": comentarios})
df_comentarios
```

### 13.2 Limpeza e pré-processamento de texto

```python
import re
import nltk
nltk.download("stopwords", quiet=True)
from nltk.corpus import stopwords

stop_pt = set(stopwords.words("portuguese"))

def limpar_texto(texto):
    texto = texto.lower()
    texto = re.sub(r"[^a-zà-ú\s]", "", texto)   # remove pontuação e números
    palavras = texto.split()
    palavras = [p for p in palavras if p not in stop_pt]
    return " ".join(palavras)

df_comentarios["texto_limpo"] = df_comentarios["comentario"].apply(limpar_texto)
df_comentarios
```

### 13.3 Vetorização de texto (Bag of Words / TF-IDF)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features=20)
X_tfidf = vectorizer.fit_transform(df_comentarios["texto_limpo"])

df_tfidf = pd.DataFrame(X_tfidf.toarray(), columns=vectorizer.get_feature_names_out())
df_tfidf.round(2)
```

### 13.4 Nuvem de palavras

```python
from wordcloud import WordCloud

texto_completo = " ".join(df_comentarios["texto_limpo"])
wc = WordCloud(width=800, height=400, background_color="white", colormap="viridis").generate(texto_completo)

plt.figure(figsize=(10, 5))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.title("Nuvem de palavras — Comentários dos alunos")
plt.show()
```

### 13.5 Análise de sentimento simples (classificador)

```python
# Rótulos manuais para exemplo didático (positivo=1, negativo=0)
sentimento = [1, 0, 1, 0, 1, 0, 1, 0]

X_train_txt, X_test_txt, y_train_txt, y_test_txt = train_test_split(
    X_tfidf, sentimento, test_size=0.25, random_state=42
)

from sklearn.naive_bayes import MultinomialNB
modelo_nb = MultinomialNB()
modelo_nb.fit(X_train_txt, y_train_txt)

print("Acurácia:", accuracy_score(y_test_txt, modelo_nb.predict(X_test_txt)).round(2))

# Testando com um comentário novo
novo_comentario = ["o curso foi maravilhoso e aprendi muito"]
novo_limpo = [limpar_texto(novo_comentario[0])]
novo_vetor = vectorizer.transform(novo_limpo)
pred = modelo_nb.predict(novo_vetor)
print("Sentimento previsto:", "Positivo" if pred[0] == 1 else "Negativo")
```

> 💡 **Aplicação real:** esse pipeline é exatamente o tipo de análise que pode ser aplicado nos comentários do canal EstaTiDados no YouTube, ou nas respostas de formulários de satisfação da CECD, para identificar automaticamente feedbacks positivos/negativos em escala.

# 14. Support Vector Machines (SVM) ⚔️

SVM busca encontrar o **hiperplano que melhor separa as classes**, maximizando a margem entre elas. Funciona muito bem em espaços de alta dimensão (como o de texto vetorizado!) e pode capturar relações não-lineares usando **kernels**.


### 14.1 SVM linear

```python
from sklearn.svm import SVC

X = df[["nota", "horas_estudo"]]
y = df["status"]

X_scaled = StandardScaler().fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.25, random_state=42, stratify=y)

svm_linear = SVC(kernel="linear", C=1.0)
svm_linear.fit(X_train, y_train)
print("Acurácia (SVM linear):", accuracy_score(y_test, svm_linear.predict(X_test)).round(2))
```

### 14.2 Kernels não-lineares (RBF)

```python
svm_rbf = SVC(kernel="rbf", C=1.0, gamma="scale")
svm_rbf.fit(X_train, y_train)
print("Acurácia (SVM RBF):", accuracy_score(y_test, svm_rbf.predict(X_test)).round(2))
```

### 14.3 Visualizando a fronteira de decisão

```python
import numpy as np

def plotar_fronteira(modelo, X, y, titulo):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.05), np.arange(y_min, y_max, 0.05))

    Z = modelo.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = pd.factorize(Z)[0].reshape(xx.shape)

    plt.figure(figsize=(7, 5))
    plt.contourf(xx, yy, Z, alpha=0.3, cmap="coolwarm")
    y_num = pd.factorize(y)[0]
    plt.scatter(X[:, 0], X[:, 1], c=y_num, cmap="coolwarm", edgecolor="k")
    plt.title(titulo)
    plt.xlabel("Nota (padronizada)")
    plt.ylabel("Horas de estudo (padronizada)")
    plt.show()

plotar_fronteira(svm_rbf, X_scaled, y, "Fronteira de decisão — SVM (kernel RBF)")
```

# 15. Sistemas de Recomendação (Netflix, YouTube, Amazon, Spotify) 🎬🎵🛒

Sistemas de recomendação sugerem itens (filmes, vídeos, produtos, músicas) com base no comportamento dos usuários. Os dois grandes enfoques são:

- **Filtragem baseada em conteúdo (content-based):** recomenda itens *parecidos* com o que o usuário já gostou (baseado em atributos do item).
- **Filtragem colaborativa (collaborative filtering):** recomenda itens que usuários *parecidos* gostaram (baseado no comportamento de outros usuários), como a Netflix e a Amazon fazem.


### 15.1 Simulando uma base de avaliações (estilo Netflix/Spotify)

```python
import numpy as np
import pandas as pd

np.random.seed(7)
usuarios = [f"user_{i}" for i in range(1, 11)]
cursos_cecd = ["Python p/ Dados", "R Básico", "Estatística I", "SQL", "Machine Learning",
               "Power BI", "Séries Temporais", "Deep Learning"]

# matriz usuário x curso, com notas de 1-5 e muitos NaN (nem todo mundo avaliou tudo)
avaliacoes = pd.DataFrame(
    np.random.choice([np.nan, 1, 2, 3, 4, 5], size=(len(usuarios), len(cursos_cecd)),
                      p=[0.5, 0.05, 0.08, 0.12, 0.15, 0.10]),
    index=usuarios, columns=cursos_cecd
)
avaliacoes
```

### 15.2 Filtragem Colaborativa baseada em Usuário (User-Based)

```python
from sklearn.metrics.pairwise import cosine_similarity

avaliacoes_preenchidas = avaliacoes.fillna(0)
similaridade_usuarios = cosine_similarity(avaliacoes_preenchidas)
sim_df = pd.DataFrame(similaridade_usuarios, index=usuarios, columns=usuarios)

def recomendar_por_usuarios_similares(usuario, n_similares=3, n_recomendacoes=3):
    similares = sim_df[usuario].drop(usuario).sort_values(ascending=False).head(n_similares)
    usuarios_similares = similares.index

    medias = avaliacoes.loc[usuarios_similares].mean().sort_values(ascending=False)
    ja_avaliados = avaliacoes.loc[usuario].dropna().index
    recomendacoes = medias.drop(labels=ja_avaliados, errors="ignore").head(n_recomendacoes)
    return recomendacoes

recomendar_por_usuarios_similares("user_1")
```

### 15.3 Filtragem Colaborativa baseada em Item (Item-Based) — como a Amazon faz

```python
similaridade_itens = cosine_similarity(avaliacoes_preenchidas.T)
sim_itens_df = pd.DataFrame(similaridade_itens, index=cursos_cecd, columns=cursos_cecd)

def cursos_parecidos(curso, n=3):
    return sim_itens_df[curso].drop(curso).sort_values(ascending=False).head(n)

print("Quem gosta de 'Python p/ Dados' também pode gostar de:")
cursos_parecidos("Python p/ Dados")
```

### 15.4 Filtragem Baseada em Conteúdo — como YouTube/Spotify recomendam por atributos

```python
cursos_atributos = pd.DataFrame({
    "curso": cursos_cecd,
    "area": ["Programação", "Estatística", "Estatística", "Programação",
             "Machine Learning", "Visualização", "Estatística", "Machine Learning"],
    "nivel": ["Básico", "Básico", "Básico", "Básico", "Avançado", "Intermediário", "Avançado", "Avançado"],
})

# One-hot encoding dos atributos + similaridade de cosseno
atributos_ohe = pd.get_dummies(cursos_atributos.set_index("curso")[["area", "nivel"]])
sim_conteudo = cosine_similarity(atributos_ohe)
sim_conteudo_df = pd.DataFrame(sim_conteudo, index=cursos_atributos["curso"], columns=cursos_atributos["curso"])

print("Alunos que gostaram de 'Machine Learning' também podem gostar de (por conteúdo):")
sim_conteudo_df["Machine Learning"].drop("Machine Learning").sort_values(ascending=False).head(3)
```

### 15.5 SVD (Fatoração de Matrizes) — a técnica que venceu o Prêmio Netflix

```python
from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(n_components=3, random_state=42)
fatores_usuarios = svd.fit_transform(avaliacoes_preenchidas)
fatores_itens = svd.components_

reconstruido = np.dot(fatores_usuarios, fatores_itens)
avaliacoes_previstas = pd.DataFrame(reconstruido, index=usuarios, columns=cursos_cecd)

# Recomendação final para um usuário: cursos com maior nota PREVISTA que ele ainda não avaliou
def recomendar_svd(usuario, n=3):
    ja_avaliados = avaliacoes.loc[usuario].dropna().index
    previsoes = avaliacoes_previstas.loc[usuario].drop(labels=ja_avaliados, errors="ignore")
    return previsoes.sort_values(ascending=False).head(n)

recomendar_svd("user_3")
```

> 💡 **No mundo real:** Netflix, YouTube, Spotify e Amazon combinam várias dessas técnicas (colaborativa + conteúdo + contexto como horário/dispositivo) em sistemas híbridos, além de redes neurais profundas para capturar padrões complexos. Mas os princípios de similaridade e fatoração de matrizes que você viu aqui são o coração de tudo isso.
